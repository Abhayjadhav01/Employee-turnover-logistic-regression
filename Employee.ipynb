```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv(r"employee_turnover.csv")
df.head()
```

```python
x=df.drop("Employee_Turnover", axis=1)
y=df["Employee_Turnover"]

x_train, x_test, y_train, y_test=train_test_split(
    x, y, test_size=0.2 , random_state=0
)

model=LogisticRegression()
model.fit(x_train, y_train)

y_pred=model.predict(x_test)

print(y_pred)
```

```python
acc_score=accuracy_score(y_test, y_pred)
print("acc_score = ", acc_score*100, "%")

precision=precision_score(y_test, y_pred)
print("precion score = ", precision*100, "%")
```

```python
#l1 regularization



model_l1=LogisticRegression(
    penalty="l1",
    solver="liblinear",
    
)

model_l1.fit(x_train, y_train)

y_pred_l1=model_l1.predict(x_test)

acc_score_l1=accuracy_score(y_test, y_pred_l1)
precision_l1=precision_score(y_test, y_pred_l1)
```

```python
#l2 regularization

model_l2=LogisticRegression(
    penalty="l2",
    solver="liblinear"
)

model_l2.fit(x_train,y_train)
y_pred_l2=model_l2.predict(x_test)

acc_score_l2=accuracy_score(y_test, y_pred_l2)
precision_l2=precision_score(y_test, y_pred_l2)
```

```python
print("acc_score = ", acc_score*100, "%", "    acc_score_l1 = ", acc_score_l1*100, "%","       acc_score_l2 = ", acc_score_l2*100, "%")
print("precion score = ", precision*100, "%","           precion score l1 = ", precision_l1*100, "%","     precion score  l2= ", precision_l2*100, "%")
```
